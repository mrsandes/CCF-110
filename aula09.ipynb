```python
# 1618

n = int(input())

for _ in range(n):
  Ax, Ay, Bx, By, Cx, Cy, Dx, Dy, Rx, Ry = map(int, input().split())
  
  if Rx >= Ax and Rx <= Bx and Ry >= Ay and Ry <= Dy:
    print(1)
      
  else:
    print(0)
```

```python
# 1245

while True:
  try: 
    N = int(input())
    botas = []
    total = 0
    
    for _ in range(N):
      atual = input()
      
      if atual.split()[1] == "E":
        if atual.split()[0] + " D" in botas:
          botas.remove(atual.split()[0] + " D")
          total += 1
            
        else:
          botas.append(atual)
              
      else:
        if atual.split()[0] + " E" in botas:
          botas.remove(atual.split()[0] + " E")
          total += 1
              
        else:
          botas.append(atual)
        
    print(total)
  
  except:
    break
```

```python
# 2534

while True:
  try:
    N, Q = map(int, input().split())

    notas = []
    for _ in range(N):
      notas.append(int(input()))

    notas.sort(reverse=True)

    for _ in range(Q):
      print(notas[int(input()) - 1])

  except:
    break
```

```python
# 2540

while True:
  try:
    N = int(input())
    
    votos = list(map(int, input().split()))
    soma = 0
    
    for i in range(N):
      soma += votos[i]
        
    if soma >= N * (2 / 3):
      print("impeachment")
        
    else:
      print("acusacao arquivada")
      
  except:
    break
```

```python
# 2542

while True:
  try:
    N = int(input())
    M, L = map(int, input().split())
    Mc, Lc = [], []
    
    for _ in range(M):
      Mc.append(input())
        
    for _ in range(L):
      Lc.append(input())
        
    Cm, Cl = map(int, input().split())
    A = int(input())
    
    if int(Mc[Cm - 1].split()[A - 1]) > int(Lc[Cl - 1].split()[A - 1]):
      print("Marcos")
        
    elif int(Mc[Cm - 1].split()[A - 1]) < int(Lc[Cl - 1].split()[A - 1]):
      print("Leonardo")
        
    else:
      print("Empate")
  
  except:
    break
```
