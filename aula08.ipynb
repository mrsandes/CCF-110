```python
# 1181

matriz = [[ 0 for _ in range(12)] for _ in range(12)]

L = int(input())
T = input()

for i in range(12):
  for j in range(12):
    matriz[i][j] = float(input())
        
soma = 0
for i in range(12):
  soma += matriz[L][i]
    
print(f"{soma:.1f}" if T == 'S' else f"{soma / 12:.1f}")
```

```python
# 1183

matriz = [[ 0  for _ in range(12)] for _ in range(12)]

O = input()

for i in range(12):
  for j in range(12):
    matriz[i][j] = float(input())
            
soma, k = 0, 0
for i in range(12):
  for j in range(12):
    if i + j > i * 2:
      soma += matriz[i][j]
      k += 1
            
print(f"{soma:.1f}" if O == 'S' else f"{soma / k:.1f}")
```

```python
# 1190

matriz = [[ 0  for _ in range(12)] for _ in range(12)]

O = input()

for i in range(12):
  for j in range(12):
    matriz[i][j] = float(input())
          
soma, k = 0, 0
for i in range(12):
  for j in range(12):
    if i + j > i * 2 and i + j > 11:
      soma += matriz[i][j]
      k += 1
          
print(f"{soma:.1f}" if O == 'S' else f"{soma / 30:.1f}")
```

```python
# 1157

while True:
  n = int(input())
  
  if n == 0:
    break
  
  e = len(str(2**((n - 1) * 2)))

  for i in range(n):
    for j in range(n):
      print(f"{2**(i + j):>{e}}", end="")
      
      if j < n - 1:
        print(" ", end="")
    print()
  print()
```
